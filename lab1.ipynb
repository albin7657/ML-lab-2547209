```python
import pandas as pd
import numpy as np

# Load datasets
aqi_df = pd.read_csv("city_day.csv")
crop_df = pd.read_csv("crop_production.csv")

print("AIR QUALITY DATASET")


print("\nShape:")
print(aqi_df.shape)

print("\nColumns:")
print(aqi_df.columns.tolist())

print("\nData Types:")
print(aqi_df.dtypes)

print("\nFirst 5 Rows:")
display(aqi_df.head())

print("\nMissing Values:")
print(aqi_df.isnull().sum())

print("\nDuplicate Records:")
print(aqi_df.duplicated().sum())

print("\nStatistical Summary:")
display(aqi_df.describe(include='all'))

print("\nUnique Values Per Column:")
for col in aqi_df.columns:
    print(f"{col}: {aqi_df[col].nunique()}")




print("CROP PRODUCTION DATASET")

print("\nShape:")
print(crop_df.shape)

print("\nColumns:")
print(crop_df.columns.tolist())

print("\nData Types:")
print(crop_df.dtypes)

print("\nFirst 5 Rows:")
display(crop_df.head())

print("\nMissing Values:")
print(crop_df.isnull().sum())

print("\nDuplicate Records:")
print(crop_df.duplicated().sum())

print("\nStatistical Summary:")
display(crop_df.describe(include='all'))

print("\nUnique Values Per Column:")
for col in crop_df.columns:
    print(f"{col}: {crop_df[col].nunique()}")
```

Before performing any analysis, it is important to understand the structure, size, and quality of the datasets. Functions such as shape, info(), describe(), and isnull() were used to obtain a complete profile of both datasets.

This approach was chosen because machine learning and analytical models depend heavily on data quality. Profiling helps identify missing values, incorrect data types, duplicate records, and potential anomalies before further processing.

During inspection, it was observed that the air quality dataset contains several missing values and the crop dataset contains information at district and state level. Since the datasets were collected differently, understanding their structure was essential before attempting any integration.

```python

aqi_clean = aqi_df.copy()

print("Missing Values BEFORE Treatment")
print(aqi_clean.isnull().sum())

# Numerical columns
num_cols = aqi_clean.select_dtypes(include=np.number).columns

for col in aqi_clean.select_dtypes(include=np.number).columns:

    missing_percent = (
        aqi_clean[col].isnull().sum() /
        len(aqi_clean)
    ) * 100

    if missing_percent < 5:
        aqi_clean[col] = aqi_clean[col].fillna(
            aqi_clean[col].mean()
        )

    else:
        aqi_clean[col] = aqi_clean[col].fillna(
            aqi_clean[col].median()
        )

print("\nMissing Values AFTER Treatment")
print(aqi_clean.isnull().sum())
```



Missing values were handled using different strategies depending on the percentage of missing data and the nature of each column. Numerical columns with moderate missing values were imputed using median values because median is less sensitive to extreme values and skewed distributions.

Rows with missing AQI values were removed because AQI is the primary variable used in subsequent analysis. Using only genuine AQI observations improves the reliability of statistical conclusions.

This approach was selected because deleting all rows would lead to excessive data loss, while replacing every missing value with a mean could distort the natural distribution of the data.

```python


crop_clean = crop_df.copy()

print("Rows Before:")
print(len(crop_clean))

crop_clean.drop_duplicates(inplace=True)

print("Rows After:")
print(len(crop_clean))

# Remove extra spaces

crop_clean['State_Name'] = (
    crop_clean['State_Name']
    .str.strip()
    .str.title()
)

# State corrections

state_mapping = {
    'Orissa':'Odisha',
    'Uttaranchal':'Uttarakhand'
}

crop_clean['State_Name'] = (
    crop_clean['State_Name']
    .replace(state_mapping)
)

print("\nUnique States:")
print(sorted(crop_clean['State_Name'].unique()))
```



Duplicate records were removed to avoid over-representation of observations. State names were cleaned by removing extra spaces and standardizing text formatting.

This approach ensures consistency during future merging operations. Since merging relies on exact matches between keys, even small differences in spelling or spaces can prevent successful integration of datasets.

Standardization improves data quality and prevents incorrect or incomplete merges.

```python
aqi_clean = aqi_clean.dropna(subset=["AQI"])

print("Remaining rows:", len(aqi_clean))
print("Missing AQI values:", aqi_clean["AQI"].isnull().sum())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10,5))

sns.histplot(
    aqi_clean["AQI"],
    bins=30,
    kde=True
)

plt.title("Distribution of AQI")
plt.xlabel("AQI")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(10,4))

sns.boxplot(
    x=aqi_clean["AQI"]
)

plt.title("AQI Outliers")
plt.show()
```

```python
Q1 = aqi_clean["AQI"].quantile(0.25)
Q3 = aqi_clean["AQI"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR

outliers = aqi_clean[
    (aqi_clean["AQI"] < lower) |
    (aqi_clean["AQI"] > upper)
]

print("Outliers:", len(outliers))
```



The Interquartile Range (IQR) method was used to identify extreme values because it is robust to skewed distributions and does not assume normality.

Instead of deleting outliers completely, extreme values were capped using Winsorization. This preserves the number of observations while reducing the influence of unrealistic values on statistical measures.

This approach was chosen because air quality measurements may naturally vary, and completely removing observations could result in the loss of meaningful information.

```python
aqi_clean["AQI"] = np.where(
    aqi_clean["AQI"] > upper,
    upper,
    aqi_clean["AQI"]
)

aqi_clean["AQI"] = np.where(
    aqi_clean["AQI"] < lower,
    lower,
    aqi_clean["AQI"]
)
```

```python
plt.figure(figsize=(10,4))

sns.boxplot(
    x=aqi_clean["AQI"]
)

plt.title("After Outlier Treatment")
plt.show()
```
