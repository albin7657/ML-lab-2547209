Before performing any analysis, it is important to understand the structure, size, and quality of the datasets. Functions such as shape, info(), describe(), and isnull() were used to obtain a complete profile of both datasets.

This approach was chosen because machine learning and analytical models depend heavily on data quality. Profiling helps identify missing values, incorrect data types, duplicate records, and potential anomalies before further processing.

During inspection, it was observed that the air quality dataset contains several missing values and the crop dataset contains information at district and state level. Since the datasets were collected differently, understanding their structure was essential before attempting any integration.

```python

aqi_clean = aqi_df.copy()

print("Missing Values BEFORE Treatment")
print(aqi_clean.isnull().sum())

# Numerical columns
num_cols = aqi_clean.select_dtypes(include=np.number).columns

# Fill all other numerical columns
for col in num_cols:
    if col != "AQI":
        missing_percent = (
            aqi_clean[col].isnull().sum() / len(aqi_clean)
        ) * 100

        if missing_percent < 5:
            aqi_clean[col] = aqi_clean[col].fillna(
                aqi_clean[col].mean()
            )
        else:
            aqi_clean[col] = aqi_clean[col].fillna(
                aqi_clean[col].median()
            )

# Remove rows with missing AQI
aqi_clean = aqi_clean.dropna(subset=["AQI"])

print("\nMissing Values AFTER Treatment")
print(aqi_clean.isnull().sum())
```

```python
aqi_clean = aqi_clean.dropna(subset=["AQI"])

print("Remaining rows:", len(aqi_clean))
print("Missing AQI values:", aqi_clean["AQI"].isnull().sum())
```

Missing values were treated differently depending on the importance of the variable and the proportion of missing data. Numerical columns with relatively few missing values were imputed using the mean, while columns with larger percentages of missing values were imputed using the median because median is more robust to skewed distributions and extreme values.

AQI values were treated separately because AQI is the primary variable used throughout the analysis. Rows with missing AQI values were removed rather than imputed to avoid introducing artificial pollution measurements that could bias trend and seasonal analyses. This strategy preserved most of the data while ensuring that conclusions were based on observed AQI values.

Before merging the datasets, several inconsistencies were investigated. Duplicate records were checked for in the crop dataset, but no duplicate rows were found. Therefore, no records needed to be removed.

The crop dataset contained trailing spaces in some state names, such as "Jammu and Kashmir " and "Telangana ". These spaces were removed using string cleaning operations to ensure consistent naming. Such inconsistencies can prevent records from matching correctly during a merge.

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10,5))

sns.histplot(
    aqi_clean["AQI"],
    bins=30,
    kde=True
)

plt.title("Distribution of AQI")
plt.xlabel("AQI")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(10,4))

sns.boxplot(
    x=aqi_clean["AQI"]
)

plt.title("AQI Outliers")
plt.show()
```

A histogram and a boxplot were used because they answer different aspects of the problem. The histogram shows how AQI values are distributed and where most observations are concentrated, making it suitable for identifying whether pollution levels are generally moderate or dominated by a few locations. A boxplot was used to detect extreme values and determine whether unusually high AQI readings could influence summary statistics such as the average.

Two important observations were made. First, most AQI values are concentrated within a moderate range, indicating that pollution levels are not uniformly severe across all cities. Second, several extreme AQI values are present, suggesting that a small number of highly polluted observations may pull the average AQI upward. Therefore, relying solely on the average may not provide a complete picture of typical air quality, and the median should also be considered when describing overall pollution levels.


```python
Q1 = aqi_clean["AQI"].quantile(0.25)
Q3 = aqi_clean["AQI"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR

outliers = aqi_clean[
    (aqi_clean["AQI"] < lower) |
    (aqi_clean["AQI"] > upper)
]

print("Outliers:", len(outliers))
```

```python
aqi_clean["AQI"] = np.where(
    aqi_clean["AQI"] > upper,
    upper,
    aqi_clean["AQI"]
)

aqi_clean["AQI"] = np.where(
    aqi_clean["AQI"] < lower,
    lower,
    aqi_clean["AQI"]
)
```

```python
plt.figure(figsize=(10,4))

sns.boxplot(
    x=aqi_clean["AQI"]
)

plt.title("After Outlier Treatment")
plt.show()
```

Extreme AQI values were identified using the Interquartile Range (IQR) method. This method was chosen because it is robust to skewed distributions and effectively detects unusually high or low observations without being strongly influenced by extreme values. Values lying below Q1 − 1.5×IQR or above Q3 + 1.5×IQR were classified as outliers.

The number of outliers was calculated to quantify the extent of the issue. Instead of removing these observations, the outlier values were capped at the lower and upper IQR limits using winsorization. This approach preserves all records while reducing the influence of unrealistic AQI readings on statistical summaries and future models.

Boxplots before and after treatment were used to visually verify the effectiveness of the procedure. After treatment, the distribution became less affected by extreme values while maintaining its overall structure. This approach was preferred because extreme values do not always represent errors, and deleting them could result in unnecessary loss of information.


```python
aqi_clean["Date"] = pd.to_datetime(
    aqi_clean["Date"],
    format="%d-%m-%Y"
)

aqi_clean["Date"] = pd.to_datetime(
    aqi_clean["Date"]
)

aqi_clean["Year"] = (
    aqi_clean["Date"].dt.year
)

yearly_aqi = (
    aqi_clean
    .groupby("Year")["AQI"]
    .mean()
    .reset_index()
)

plt.figure(figsize=(10,5))

sns.lineplot(
    data=yearly_aqi,
    x="Year",
    y="AQI",
    marker="o"
)

plt.title("Average AQI by Year")
plt.show()
```

```python
max_row = yearly_aqi.loc[yearly_aqi["AQI"].idxmax()]
min_row = yearly_aqi.loc[yearly_aqi["AQI"].idxmin()]

plt.figure(figsize=(10,5))

sns.lineplot(
    data=yearly_aqi,
    x="Year",
    y="AQI",
    marker="o"
)

# Highlight highest AQI year
plt.scatter(
    max_row["Year"],
    max_row["AQI"],
    color="red",
    s=100,
    label="Most Polluted Year"
)

# Highlight lowest AQI year
plt.scatter(
    min_row["Year"],
    min_row["AQI"],
    color="green",
    s=100,
    label="Least Polluted Year"
)

plt.title("Average AQI by Year")
plt.xlabel("Year")
plt.ylabel("Average AQI")
plt.legend()

plt.show()
```

```python
print("Most populated year:",
    yearly_aqi.loc[
        yearly_aqi["AQI"].idxmax()
    ]
)

print("Least polluted year:",
    yearly_aqi.loc[
        yearly_aqi["AQI"].idxmin()
    ]
)
```

Based on the yearly average AQI values, air quality has fluctuated over the years rather than showing a steady improvement. The most polluted year recorded the highest average AQI, while the least polluted year recorded the lowest average AQI. Although some years show lower pollution levels, the data does not indicate a clear and consistent improvement after 2018. Therefore, the available evidence suggests that air quality has varied over time rather than continuously improving or worsening. These results indicate that while pollution control measures may have had some impact, the overall trend is not strong enough to conclude that air quality has steadily improved.

Monthly average AQI values were analysed to investigate whether air quality follows a seasonal pattern. Monthly aggregation was chosen because it provides sufficient detail to reveal recurring variations throughout the year while avoiding the noise present in daily observations. A line plot was used to make these changes easy to visualize.

The results indicate that AQI varies across months, confirming the presence of seasonality. In particular, pollution levels increase during the October–December period, which corresponds to the harvest season. This finding supports the agricultural NGO's claim that air quality tends to worsen during these months. However, crop residue burning may not be the only reason for this increase. Weather conditions, temperature changes, and other human activities may also contribute to the observed pattern. Therefore, the analysis supports the claim that pollution is higher during the harvest season, but it does not prove that crop residue burning is the sole cause.

The two datasets could not be merged directly because they were collected at different levels of detail. The air quality dataset contained daily observations for individual cities, whereas the crop dataset contained yearly records. To make them compatible, AQI values were aggregated to the state-year level by averaging observations belonging to the same state and year. Crop records were also aggregated to the state-year level by summing cultivated area and production values, and the year columns were standardized to create common merge keys.

After merging the datasets, a correlation matrix was used to investigate relationships among AQI, cultivated area, and crop production. However, the resulting heatmap contained missing values. Additional diagnostic checks showed that only a small number of states and years overlapped between the two datasets, resulting in insufficient matching records after the merge.

Therefore, no reliable relationship between air quality and crop production could be established.


First, air quality has fluctuated over the years rather than showing a clear and steady improvement, indicating that pollution remains a continuing concern. Second, pollution levels follow a seasonal pattern, with air quality tending to worsen during the October to December period. This suggests that certain seasonal activities and weather conditions contribute to higher pollution levels. Third, a direct relationship between air quality and crop production could not be established because the available datasets had limited overlap in terms of states and years.

For farmers, these findings mean that poor air quality may coincide with important agricultural periods, but the current evidence is not strong enough to conclude that pollution directly reduces crop production.

Based on these results, I recommend strengthening pollution-control measures during periods when air quality is consistently poor and promoting cleaner agricultural practices.

One important limitation is that the data does not account for factors such as rainfall, irrigation, soil quality, and farming practices, so it cannot prove cause and effect.
