```python
# ============================================================
# CREATE FIRE EVENT PRIORITY TARGET
# ============================================================

# The 75th percentile was identified during the FRP distribution audit
FRP_PRIORITY_THRESHOLD = 25.1

df["high_priority"] = (
    df["frp"] >= FRP_PRIORITY_THRESHOLD
).astype(int)

print("Priority threshold (FRP):", FRP_PRIORITY_THRESHOLD)

print("\nTARGET DISTRIBUTION:")
print(df["high_priority"].value_counts())

print("\nTARGET DISTRIBUTION (%):")
print(
    (df["high_priority"]
     .value_counts(normalize=True) * 100)
    .round(2)
)
```

```python
# ============================================================
# TARGET CLASS DISTRIBUTION
# ============================================================

class_counts = df["high_priority"].value_counts().sort_index()

plt.figure(figsize=(7, 5))
plt.bar(
    ["Routine Priority (0)", "High Priority (1)"],
    class_counts.values
)

plt.xlabel("Fire Event Priority")
plt.ylabel("Number of Events")
plt.title("Distribution of Fire Event Priority Classes")
plt.show()
```

```python
# ============================================================
# DATA TYPE CONVERSION AND TEMPORAL FEATURE ENGINEERING
# ============================================================

# Work on a copy
data = df.copy()

# Remove the exploratory target created earlier
if "high_priority" in data.columns:
    data = data.drop(columns=["high_priority"])

# Convert acquisition date to datetime
data["acq_date"] = pd.to_datetime(data["acq_date"])

# Create temporal features
data["year"] = data["acq_date"].dt.year
data["month"] = data["acq_date"].dt.month
data["day"] = data["acq_date"].dt.day
data["day_of_week"] = data["acq_date"].dt.dayofweek

# Extract hour and minute from acquisition time
data["acq_time"] = data["acq_time"].astype(str).str.zfill(4)

data["hour"] = data["acq_time"].str[:2].astype(int)
data["minute"] = data["acq_time"].str[2:].astype(int)

print("Shape after feature engineering:", data.shape)

display(
    data[
        [
            "acq_date", "acq_time",
            "year", "month", "day",
            "day_of_week", "hour", "minute"
        ]
    ].head()
)
```

```python
# ============================================================
# REPRODUCIBLE TRAIN / VALIDATION / TEST SPLIT
# ============================================================

from sklearn.model_selection import train_test_split

# First split:
# 80% development data
# 20% untouched test data
train_val_data, test_data = train_test_split(
    data,
    test_size=0.20,
    random_state=RANDOM_STATE,
    shuffle=True
)

# Split the 80% development data into:
# 70% training overall
# 10% validation overall
train_data, val_data = train_test_split(
    train_val_data,
    test_size=0.125,
    random_state=RANDOM_STATE,
    shuffle=True
)

print("Training set:", train_data.shape)
print("Validation set:", val_data.shape)
print("Test set:", test_data.shape)

print("\nSplit percentages:")
print("Train:", round(len(train_data) / len(data) * 100, 2), "%")
print("Validation:", round(len(val_data) / len(data) * 100, 2), "%")
print("Test:", round(len(test_data) / len(data) * 100, 2), "%")
```

```python
# ============================================================
# LEAKAGE-SAFE TARGET CREATION
# ============================================================

# Learn the priority threshold ONLY from training data
FRP_PRIORITY_THRESHOLD = train_data["frp"].quantile(0.75)

print(
    "Training-data 75th percentile FRP threshold:",
    round(FRP_PRIORITY_THRESHOLD, 3)
)

# Apply the SAME threshold to all three datasets
for dataset in [train_data, val_data, test_data]:
    dataset["high_priority"] = (
        dataset["frp"] >= FRP_PRIORITY_THRESHOLD
    ).astype(int)

print("\nTRAIN TARGET DISTRIBUTION:")
display(
    train_data["high_priority"]
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(2)
)

print("\nVALIDATION TARGET DISTRIBUTION:")
display(
    val_data["high_priority"]
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(2)
)

print("\nTEST TARGET DISTRIBUTION:")
display(
    test_data["high_priority"]
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(2)
)
```

```python
# ============================================================
# DEFINE FINAL MODEL FEATURES
# ============================================================

drop_columns = [
    "frp",            # Used to define target -> direct leakage
    "high_priority",  # Target
    "acq_date",       # Replaced by engineered temporal features
    "acq_time",       # Replaced by hour and minute
    "instrument"      # Constant column: MODIS for all records
]

feature_columns = [
    col for col in train_data.columns
    if col not in drop_columns
]

print("Final feature columns:")
print(feature_columns)

print("\nNumber of features:", len(feature_columns))
```

```python
# ============================================================
# CHECK FOR CONSTANT FEATURES
# ============================================================

constant_columns = []

for col in feature_columns:
    if train_data[col].nunique() <= 1:
        constant_columns.append(col)

print("Constant columns found:", constant_columns)
```

```python
# ============================================================
# CREATE TRAIN, VALIDATION AND TEST FEATURES
# ============================================================

X_train = train_data[feature_columns].copy()
y_train = train_data["high_priority"].copy()

X_val = val_data[feature_columns].copy()
y_val = val_data["high_priority"].copy()

X_test = test_data[feature_columns].copy()
y_test = test_data["high_priority"].copy()

print("X_train shape:", X_train.shape)
print("X_val shape:", X_val.shape)
print("X_test shape:", X_test.shape)

print("\ny_train distribution:")
print(y_train.value_counts(normalize=True).sort_index().round(4))
```

```python
# ============================================================
# REPRODUCIBLE STRATIFIED TRAINING SAMPLE
# ============================================================

TRAIN_SAMPLE_SIZE = 120_000

if len(X_train) > TRAIN_SAMPLE_SIZE:
    
    X_train_sample, _, y_train_sample, _ = train_test_split(
        X_train,
        y_train,
        train_size=TRAIN_SAMPLE_SIZE,
        stratify=y_train,
        random_state=RANDOM_STATE
    )
    
else:
    X_train_sample = X_train.copy()
    y_train_sample = y_train.copy()

print("Original training size:", len(X_train))
print("Training sample size:", len(X_train_sample))

print("\nSample target distribution:")
print(
    y_train_sample
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(2)
)
```

```python
# ============================================================
# IDENTIFY NUMERICAL AND CATEGORICAL FEATURES
# ============================================================

numerical_features = X_train_sample.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X_train_sample.select_dtypes(
    include=["object", "category"]
).columns.tolist()

print("NUMERICAL FEATURES:")
print(numerical_features)

print("\nCATEGORICAL FEATURES:")
print(categorical_features)
```

```python
# ============================================================
# LEAKAGE-SAFE PREPROCESSING PIPELINE
# ============================================================

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, RobustScaler

numerical_transformer = Pipeline(
    steps=[
        ("scaler", RobustScaler())
    ]
)

categorical_transformer = Pipeline(
    steps=[
        ("onehot", OneHotEncoder(
            handle_unknown="ignore"
        ))
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numerical_transformer, numerical_features),
        ("cat", categorical_transformer, categorical_features)
    ],
    remainder="drop"
)

print("Leakage-safe preprocessing pipeline created successfully.")
```

## Preprocessing Decisions and Interpretation

### Missing Values
All 15 original variables were audited for missing and NaN values. No missing values were found, so no imputation or row deletion was required. Applying artificial imputation when no values are missing would unnecessarily modify the data.

### Duplicates
No exact duplicate rows were found. Therefore, no records were removed during duplicate cleaning.

### Invalid Values
Latitude and longitude were checked against valid geographic ranges, and relevant numerical variables were checked for invalid negative values. No invalid values were identified in the checks performed.

### Outliers
FRP was highly right-skewed. Extreme FRP values were retained because, in this domain, unusually large values may represent meaningful high-intensity fire events rather than data errors. A log transformation was used only for exploratory visualization to better understand the distribution.

### Encoding and Scaling
Categorical variables were handled using one-hot encoding. Numerical variables were scaled using RobustScaler within the preprocessing pipeline. Keeping preprocessing inside the pipeline ensures that transformations are learned from training data rather than from validation or test data.

### Target Leakage Prevention
FRP was used to derive the High-Priority target but was excluded from the model input features. Including FRP as a predictor would allow the model to directly learn the rule used to create the target and would produce misleadingly optimistic performance.

```python
# ============================================================
# TEST PREPROCESSING
# ============================================================

X_train_transformed = preprocessor.fit_transform(X_train_sample)

print("Original training sample shape:", X_train_sample.shape)
print("Transformed training sample shape:", X_train_transformed.shape)
```

```python
# ============================================================
# CROSS-VALIDATION STRATEGY
# ============================================================

from sklearn.model_selection import StratifiedKFold, cross_validate

cv = StratifiedKFold(
    n_splits=3,
    shuffle=True,
    random_state=RANDOM_STATE
)

scoring = {
    "roc_auc": "roc_auc",
    "f1": "f1",
    "precision": "precision",
    "recall": "recall"
}

print("3-fold Stratified Cross-Validation configured.")
```

```python
# ============================================================
# MODEL EVALUATION FUNCTION
# ============================================================

def evaluate_cv(model, X, y, model_name):
    
    results = cross_validate(
        model,
        X,
        y,
        cv=cv,
        scoring=scoring,
        n_jobs=-1,
        return_train_score=False
    )
    
    summary = {
        "Model": model_name,
        "CV ROC-AUC": results["test_roc_auc"].mean(),
        "CV F1": results["test_f1"].mean(),
        "CV Precision": results["test_precision"].mean(),
        "CV Recall": results["test_recall"].mean()
    }
    
    print(f"\n{model_name}")
    print("-" * 50)
    
    for metric, value in summary.items():
        if metric != "Model":
            print(f"{metric}: {value:.4f}")
    
    return summary
```

```python
# ============================================================
# DECISION TREE BASELINE
# ============================================================

from sklearn.tree import DecisionTreeClassifier

decision_tree_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            DecisionTreeClassifier(
                random_state=RANDOM_STATE,
                class_weight="balanced"
            )
        )
    ]
)

dt_results = evaluate_cv(
    decision_tree_pipeline,
    X_train_sample,
    y_train_sample,
    "Decision Tree (Baseline)"
)
```

```python
# ============================================================
# RANDOM FOREST - BAGGING MODEL
# ============================================================

from sklearn.ensemble import RandomForestClassifier

random_forest_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            RandomForestClassifier(
                n_estimators=200,
                max_depth=20,
                min_samples_split=10,
                min_samples_leaf=5,
                class_weight="balanced",
                random_state=RANDOM_STATE,
                n_jobs=-1
            )
        )
    ]
)

rf_results = evaluate_cv(
    random_forest_pipeline,
    X_train_sample,
    y_train_sample,
    "Random Forest (Bagging)"
)
```

```python
# ============================================================
# ADABOOST - BOOSTING MODEL
# ============================================================

from sklearn.ensemble import AdaBoostClassifier

adaboost_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            AdaBoostClassifier(
                n_estimators=150,
                learning_rate=0.8,
                random_state=RANDOM_STATE
            )
        )
    ]
)

ada_results = evaluate_cv(
    adaboost_pipeline,
    X_train_sample,
    y_train_sample,
    "AdaBoost (Boosting)"
)
```

```python
# ============================================================
# COMPARE INITIAL CV RESULTS
# ============================================================

cv_results = pd.DataFrame([
    dt_results,
    rf_results,
    ada_results
])

cv_results = cv_results.sort_values(
    by="CV F1",
    ascending=False
).reset_index(drop=True)

display(cv_results)
```

```python
# ============================================================
# HETEROGENEOUS STACKING CLASSIFIER
# ============================================================

from sklearn.ensemble import StackingClassifier
from sklearn.linear_model import LogisticRegression

stacking_model = StackingClassifier(
    estimators=[
        (
            "decision_tree",
            DecisionTreeClassifier(
                max_depth=15,
                min_samples_split=10,
                min_samples_leaf=5,
                class_weight="balanced",
                random_state=RANDOM_STATE
            )
        ),
        (
            "random_forest",
            RandomForestClassifier(
                n_estimators=100,
                max_depth=15,
                min_samples_split=10,
                min_samples_leaf=5,
                class_weight="balanced",
                random_state=RANDOM_STATE,
                n_jobs=-1
            )
        ),
        (
            "adaboost",
            AdaBoostClassifier(
                n_estimators=100,
                learning_rate=0.8,
                random_state=RANDOM_STATE
            )
        )
    ],
    
    final_estimator=LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        random_state=RANDOM_STATE
    ),
    
    cv=3,
    stack_method="predict_proba",
    n_jobs=-1
)
```

```python
# ============================================================
# STACKING PIPELINE
# ============================================================

stacking_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", stacking_model)
    ]
)

print("Heterogeneous stacking pipeline created successfully.")
```

```python
# ============================================================
# CROSS-VALIDATE STACKING MODEL
# ============================================================

stacking_results = evaluate_cv(
    stacking_pipeline,
    X_train_sample,
    y_train_sample,
    "Stacking Ensemble"
)
```

```python
# ============================================================
# COMPLETE CROSS-VALIDATION COMPARISON
# ============================================================

cv_results = pd.DataFrame([
    dt_results,
    rf_results,
    ada_results,
    stacking_results
])

cv_results = cv_results.sort_values(
    by="CV F1",
    ascending=False
).reset_index(drop=True)

display(cv_results)
```

```python
# ============================================================
# RANDOM FOREST TUNING
# ============================================================

from sklearn.model_selection import RandomizedSearchCV

rf_param_grid = {
    "model__n_estimators": [150, 200, 250],
    "model__max_depth": [15, 20, 25, None],
    "model__min_samples_split": [5, 10, 20],
    "model__min_samples_leaf": [2, 5, 10]
}

rf_search = RandomizedSearchCV(
    estimator=random_forest_pipeline,
    param_distributions=rf_param_grid,
    n_iter=8,
    scoring="f1",
    cv=cv,
    n_jobs=-1,
    random_state=RANDOM_STATE,
    verbose=1
)

rf_search.fit(X_train_sample, y_train_sample)

print("Best RF parameters:")
print(rf_search.best_params_)

print("\nBest CV F1:")
print(round(rf_search.best_score_, 4))
```

```python
# ============================================================
# ADABOOST TUNING
# ============================================================

ada_param_grid = {
    "model__n_estimators": [100, 150, 200, 250],
    "model__learning_rate": [0.1, 0.5, 0.8, 1.0]
}

ada_search = RandomizedSearchCV(
    estimator=adaboost_pipeline,
    param_distributions=ada_param_grid,
    n_iter=6,
    scoring="f1",
    cv=cv,
    n_jobs=-1,
    random_state=RANDOM_STATE,
    verbose=1
)

ada_search.fit(X_train_sample, y_train_sample)

print("Best AdaBoost parameters:")
print(ada_search.best_params_)

print("\nBest CV F1:")
print(round(ada_search.best_score_, 4))
```

```python
# ============================================================
# RETRIEVE TUNED MODELS
# ============================================================

best_rf = rf_search.best_estimator_
best_ada = ada_search.best_estimator_

print("Best Random Forest and AdaBoost models loaded.")
```

```python
# ============================================================
# VALIDATION EVALUATION FUNCTION
# ============================================================

from sklearn.metrics import (
    roc_auc_score,
    f1_score,
    precision_score,
    recall_score
)

def evaluate_validation(model, X, y, model_name):
    
    y_pred = model.predict(X)
    y_prob = model.predict_proba(X)[:, 1]
    
    return {
        "Model": model_name,
        "ROC-AUC": roc_auc_score(y, y_prob),
        "F1": f1_score(y, y_pred),
        "Precision": precision_score(y, y_pred),
        "Recall": recall_score(y, y_pred)
    }
```

```python
# ============================================================
# FIT INITIAL MODELS ON TRAINING SAMPLE
# ============================================================

decision_tree_pipeline.fit(X_train_sample, y_train_sample)
stacking_pipeline.fit(X_train_sample, y_train_sample)

print("Decision Tree and Stacking fitted.")
```

```python
# ============================================================
# VALIDATION MODEL COMPARISON
# ============================================================

validation_results = pd.DataFrame([
    evaluate_validation(
        decision_tree_pipeline,
        X_val,
        y_val,
        "Decision Tree (Baseline)"
    ),
    
    evaluate_validation(
        best_rf,
        X_val,
        y_val,
        "Random Forest (Tuned)"
    ),
    
    evaluate_validation(
        best_ada,
        X_val,
        y_val,
        "AdaBoost (Tuned)"
    ),
    
    evaluate_validation(
        stacking_pipeline,
        X_val,
        y_val,
        "Stacking Ensemble"
    )
])

validation_results = validation_results.sort_values(
    by="F1",
    ascending=False
).reset_index(drop=True)

display(validation_results)
```

```python
# ============================================================
# MODEL COMPARISON VISUALIZATION
# ============================================================

metrics_to_plot = ["ROC-AUC", "F1", "Precision", "Recall"]

validation_results_plot = validation_results.set_index("Model")

ax = validation_results_plot[metrics_to_plot].plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Validation Performance Comparison of ML Models")
plt.ylabel("Score")
plt.xlabel("Model")
plt.ylim(0, 1)
plt.xticks(rotation=20)
plt.legend(title="Metric")
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# SELECT BEST MODEL
# ============================================================

best_model_name = validation_results.iloc[0]["Model"]

print("Best model based on validation F1-score:")
print(best_model_name)
```

```python
# ============================================================
# SELECT THE FINAL MODEL OBJECT
# ============================================================

model_map = {
    "Decision Tree (Baseline)": decision_tree_pipeline,
    "Random Forest (Tuned)": best_rf,
    "AdaBoost (Tuned)": best_ada,
    "Stacking Ensemble": stacking_pipeline
}

final_model = model_map[best_model_name]

print("Final selected model:", best_model_name)
```

```python
# ============================================================
# FINAL UNTOUCHED TEST SET EVALUATION
# ============================================================

from sklearn.metrics import (
    roc_auc_score,
    f1_score,
    precision_score,
    recall_score,
    accuracy_score,
    classification_report
)

# Predictions
y_test_pred = final_model.predict(X_test)
y_test_prob = final_model.predict_proba(X_test)[:, 1]

# Metrics
test_metrics = {
    "Selected Model": best_model_name,
    "Accuracy": accuracy_score(y_test, y_test_pred),
    "ROC-AUC": roc_auc_score(y_test, y_test_prob),
    "F1": f1_score(y_test, y_test_pred),
    "Precision": precision_score(y_test, y_test_pred),
    "Recall": recall_score(y_test, y_test_pred)
}

print("FINAL TEST SET RESULTS")
print("=" * 50)

for metric, value in test_metrics.items():
    if metric != "Selected Model":
        print(f"{metric}: {value:.4f}")

print("\nSelected Model:")
print(test_metrics["Selected Model"])

print("\nCLASSIFICATION REPORT:")
print(
    classification_report(
        y_test,
        y_test_pred,
        target_names=["Routine Priority", "High Priority"]
    )
)
```

```python
# ============================================================
# CONFUSION MATRIX
# ============================================================

from sklearn.metrics import ConfusionMatrixDisplay, confusion_matrix

cm = confusion_matrix(y_test, y_test_pred)

print("Confusion Matrix:")
print(cm)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Routine Priority", "High Priority"]
)

disp.plot()
plt.title(f"Confusion Matrix — {best_model_name}")
plt.show()
```

# Model Ethics and Responsible Use

## Bias and Fairness
The dataset represents satellite-detected thermal anomalies rather than a complete record of all fire events. Detection quality may vary across locations, satellite overpass times, cloud conditions, land cover, and other environmental factors. Therefore, model performance may not be uniform across all regions of India. Before operational deployment, performance should be evaluated separately across geographic regions and relevant environmental conditions.

## Privacy
The NASA satellite dataset used in this project does not contain directly identifiable personal information. However, predictions should not be linked to individuals or used for surveillance purposes.

## Uncertainty
A predicted probability represents the model's estimate based on patterns in historical data and should not be interpreted as certainty. Predictions near the decision threshold require additional human verification.

## False Positives
A false positive occurs when a routine event is incorrectly classified as High Priority. This may lead to unnecessary allocation of monitoring or response resources.

## False Negatives
A false negative occurs when a genuinely High-Priority event is classified as Routine Priority. This could delay attention to an important event and may have greater environmental consequences.

## Human Oversight
The system should be used as a decision-support tool. Environmental agencies or trained personnel should verify high-priority predictions using additional information such as local reports, weather conditions, ground observations, and other satellite sources.

## Deployment Limitations
This model was trained using historical NASA MODIS observations from India and may not generalize equally to other countries, sensors, future environmental conditions, or unseen fire patterns. The High-Priority label is also an operational proxy derived from historical FRP values rather than an official emergency classification.
