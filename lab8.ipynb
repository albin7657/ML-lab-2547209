### Import Required Libraries

In this step, we import all the necessary Python libraries required for data preprocessing, model training, evaluation, and comparison.

```python
import pandas as pd

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split

from sklearn.naive_bayes import CategoricalNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
import warnings
warnings.filterwarnings("ignore")


from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)
```

## Load the Play Tennis Dataset

Load the Play Tennis dataset into a Pandas DataFrame and display the dataset for verification.

```python
data = pd.read_csv("play_tennis.csv")

print("Original Dataset:")
data
```

## Separate Input Features and Target Variable

Separate the predictor variables (Outlook, Temperature, Humidity, Wind) from the target variable (Play).

```python
X = data[['Outlook', 'Temperature', 'Humidity', 'Wind']]
y = data['Play Tennis']

print("Features:")
display(X)

print("Target:")
display(y)
```

## Data Preprocessing – Encode Categorical Values

Convert all categorical feature values and target labels into numerical values using Label Encoding so that they can be processed by machine learning algorithms.

```python
feature_encoders = {}

for column in X.columns:
    encoder = LabelEncoder()
    X[column] = encoder.fit_transform(X[column])
    feature_encoders[column] = encoder

target_encoder = LabelEncoder()
y = target_encoder.fit_transform(y)

print("Encoded Features:")
display(X)

print("Encoded Target:")
print(y)
```

## Dataset Partitioning

Split the encoded dataset into training and testing subsets using an 80:20 train-test ratio.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training Samples :", len(X_train))
print("Testing Samples  :", len(X_test))
```

## Train the Categorical Naive Bayes Classifier

Create and train a Categorical Naive Bayes (CategoricalNB) classifier using the training dataset.

```python
nb_model = CategoricalNB()

nb_model.fit(X_train, y_train)

print("Categorical Naive Bayes Model Trained Successfully.")
```

## Predict the Test Dataset

Use the trained Categorical Naive Bayes model to predict the class labels for the testing dataset.

```python
y_pred = nb_model.predict(X_test)

print("Predicted Labels:")
print(y_pred)
```

## Evaluate the Naive Bayes Model

Evaluate the performance of the trained model by calculating:

- Model Accuracy
- Confusion Matrix
- Classification Report (Precision, Recall, F1-Score)

```python
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:")
print(accuracy)

print("\nConfusion Matrix")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report")
print(classification_report(y_test, y_pred))
```

## Single-Sample Prediction

Predict whether a person will play tennis for the following weather conditions:

- Outlook: Sunny
- Temperature: Cool
- Humidity: High
- Wind: Strong

Display both the predicted class label and the corresponding class probabilities.

```python
sample = {
    "Outlook": "Sunny",
    "Temperature": "Cool",
    "Humidity": "High",
    "Wind": "Strong"
}

sample_encoded = [[
    feature_encoders["Outlook"].transform([sample["Outlook"]])[0],
    feature_encoders["Temperature"].transform([sample["Temperature"]])[0],
    feature_encoders["Humidity"].transform([sample["Humidity"]])[0],
    feature_encoders["Wind"].transform([sample["Wind"]])[0]
]]

prediction = nb_model.predict(sample_encoded)
probability = nb_model.predict_proba(sample_encoded)

print("Input Weather:")
print(sample)

print("\nPrediction:")
print(target_encoder.inverse_transform(prediction)[0])

print("\nClass Probabilities:")

for label, prob in zip(target_encoder.classes_, probability[0]):
    print(label, ":", round(prob,4))
```

## Train the Decision Tree Classifier

Train a Decision Tree classifier using the same training dataset and evaluate its prediction performance.

```python
dt_model = DecisionTreeClassifier(random_state=42)

dt_model.fit(X_train, y_train)

dt_prediction = dt_model.predict(X_test)

dt_accuracy = accuracy_score(y_test, dt_prediction)

dt_sample_prediction = dt_model.predict(sample_encoded)

dt_probability = dt_model.predict_proba(sample_encoded)

print("Decision Tree Accuracy:", dt_accuracy)
```

## Train the Logistic Regression Classifier

Train a Logistic Regression classifier using the same training dataset and evaluate its prediction performance.

```python
lr_model = LogisticRegression(max_iter=500)

lr_model.fit(X_train, y_train)

lr_prediction = lr_model.predict(X_test)

lr_accuracy = accuracy_score(y_test, lr_prediction)

lr_sample_prediction = lr_model.predict(sample_encoded)

lr_probability = lr_model.predict_proba(sample_encoded)

print("Logistic Regression Accuracy:", lr_accuracy)
```

## Train the Support Vector Machine (SVM)

Train a Support Vector Machine classifier using the same training dataset and evaluate its prediction performance.

```python
svm_model = SVC(probability=True)

svm_model.fit(X_train, y_train)

svm_prediction = svm_model.predict(X_test)

svm_accuracy = accuracy_score(y_test, svm_prediction)

svm_sample_prediction = svm_model.predict(sample_encoded)

svm_probability = svm_model.predict_proba(sample_encoded)

print("SVM Accuracy:", svm_accuracy)
```

## Compare All Models

Compare the performance of the following classifiers:

- Categorical Naive Bayes
- Decision Tree
- Logistic Regression
- Support Vector Machine (SVM)

Comparison is based on:

- Test Accuracy
- Single-Sample Prediction

```python
comparison = pd.DataFrame({
    "Model": [
        "Categorical Naive Bayes",
        "Decision Tree",
        "Logistic Regression",
        "Support Vector Machine"
    ],
    "Accuracy": [
        accuracy,
        dt_accuracy,
        lr_accuracy,
        svm_accuracy
    ],
    "Prediction": [
        target_encoder.inverse_transform(prediction)[0],
        target_encoder.inverse_transform(dt_sample_prediction)[0],
        target_encoder.inverse_transform(lr_sample_prediction)[0],
        target_encoder.inverse_transform(svm_sample_prediction)[0]
    ]
})

comparison
```

## Compare Class Probabilities

Compare the predicted class probabilities generated by all four machine learning models for the given weather conditions.

```python
print("Categorical Naive Bayes")
print(nb_model.predict_proba(sample_encoded))

print("\nDecision Tree")
print(dt_probability)

print("\nLogistic Regression")
print(lr_probability)

print("\nSupport Vector Machine")
print(svm_probability)
```
